# 3. GF(2^8) GALOIS FIELD ARITHMETIC

```python
def g_f_add(a, b):
    """Addition in Galois Field GF(2^8) using XOR."""
    return a ^ b

def xtime(byte):
    """Multiplies a single byte by x (0x02) in GF(2^8)."""
    if byte & MSB_MASK:
        shifted = byte << 1
        return (shifted ^ POLY_MODULUS) & 0xFF
    return (byte << 1) & 0xFF

def g_f_mult(a, b):
    """Multiplication of two bytes in Galois Field GF(2^8)."""
    result = 0
    current_a = a
    for _ in range(8):
        if b & 1:
            result ^= current_a
        current_a = xtime(current_a)
        b >>= 1
    return result
```

# 4. KEY EXPANSION FUNCTIONS

```python
def rot_word(word):
    """Cyclically shifts 4 bytes of a word to the left."""
    return word[1:] + word[:1]

def sub_word(word):
    """Applies S-Box substitution to each of the 4 bytes in a word."""
    new_word = []
    for byte in word:
        new_word.append(S_BOX[byte])
    return new_word

def add_rcon(word, round_num):
    """XORs the first byte of a word with the corresponding Rcon constant."""
    rcon_value = Rcon[round_num - 1]
    new_word = list(word)
    new_word[0] ^= rcon_value
    return new_word

def key_expansion(key):
    """Generates the Key Schedule for AES-128."""
    word_list = []

    for i in range(4):
        word_list.append(list(key[i*4:(i+1)*4]))

    for i in range(4, 4 * (NR + 1)):
        temp_word = list(word_list[i-1])

        if i % 4 == 0:
            temp_word = rot_word(temp_word)
            temp_word = sub_word(temp_word)

            round_num_index = i // 4
            temp_word = add_rcon(temp_word, round_num_index)

        new_word = []
        for j in range(4):
            new_word.append(word_list[i-4][j] ^ temp_word[j])

        word_list.append(new_word)

    return word_list
```

# 5. STATE CONVERSION FUNCTIONS (Block <-> Matrix)

```python
def bytes_to_state(data_chunk):
    """Converts a 16-byte block into a 4x4 state matrix (column-major order)."""
    state = [[0] * 4 for _ in range(4)]
    for i in range(16):
        byte = data_chunk[i]
        col = i // 4
        row = i % 4
        state[row][col] = byte
    return state

def state_to_bytes(state):
    """Converts a 4x4 state matrix back into a 16-byte block (column-major order)."""
    data_chunk = [0] * 16
    for col in range(4):
        for row in range(4):
            data_chunk[col * 4 + row] = state[row][col]
    return data_chunk
```

# 6. ENCRYPTION TRANSFORMATIONS

```python
def sub_bytes(state):
    """Applies the SubBytes transformation to the 4x4 state matrix."""
    for r in range(4):
        for c in range(4):
            state[r][c] = S_BOX[state[r][c]]
    return state

def shift_rows(state):
    """Cyclically shifts the rows of the state matrix to the left (by 0, 1, 2, and 3 positions)."""
    for r in range(4):
        if r == 1:
            state[r] = state[r][1:] + state[r][:1]
        elif r == 2:
            state[r] = state[r][2:] + state[r][:2]
        elif r == 3:
            state[r] = state[r][3:] + state[r][:3]
    return state

def mix_columns(state):
    """Applies the MixColumns transformation to the state matrix."""
    for c in range(4):
        col_copy = []
        for r in range(4):
            col_copy.append(state[r][c])

        state[0][c] = (
            g_f_mult(MIX_COL_MATRIX[0][0], col_copy[0]) ^
            g_f_mult(MIX_COL_MATRIX[0][1], col_copy[1]) ^
            g_f_mult(MIX_COL_MATRIX[0][2], col_copy[2]) ^
            g_f_mult(MIX_COL_MATRIX[0][3], col_copy[3])
        )
        state[1][c] = (
            g_f_mult(MIX_COL_MATRIX[1][0], col_copy[0]) ^
            g_f_mult(MIX_COL_MATRIX[1][1], col_copy[1]) ^
            g_f_mult(MIX_COL_MATRIX[1][2], col_copy[2]) ^
            g_f_mult(MIX_COL_MATRIX[1][3], col_copy[3])
        )
        state[2][c] = (
            g_f_mult(MIX_COL_MATRIX[2][0], col_copy[0]) ^
            g_f_mult(MIX_COL_MATRIX[2][1], col_copy[1]) ^
            g_f_mult(MIX_COL_MATRIX[2][2], col_copy[2]) ^
            g_f_mult(MIX_COL_MATRIX[2][3], col_copy[3])
        )
        state[3][c] = (
            g_f_mult(MIX_COL_MATRIX[3][0], col_copy[0]) ^
            g_f_mult(MIX_COL_MATRIX[3][1], col_copy[1]) ^
            g_f_mult(MIX_COL_MATRIX[3][2], col_copy[2]) ^
            g_f_mult(MIX_COL_MATRIX[3][3], col_copy[3])
        )
    return state

def add_round_key(state, round_key_words):
    """Applies the AddRoundKey operation (XOR) to the state matrix."""
    for r in range(4):
        for c in range(4):
            state[r][c] = state[r][c] ^ round_key_words[c][r]
    return state
```

# 7. DECRYPTION TRANSFORMATIONS

```python
def inv_sub_bytes(state):
    """Applies the inverse SubBytes transformation to the 4x4 state matrix."""
    for r in range(4):
        for c in range(4):
            state[r][c] = INV_S_BOX[state[r][c]]
    return state

def inv_shift_rows(state):
    """Cyclically shifts the rows of the state matrix to the right (by 0, 1, 2, and 3 positions)."""
    for r in range(4):
        if r == 1:
            state[r] = state[r][-1:] + state[r][:-1]
        elif r == 2:
            state[r] = state[r][-2:] + state[r][:-2]
        elif r == 3:
            state[r] = state[r][-3:] + state[r][:-3]
    return state

def inv_mix_columns(state):
    """Applies the inverse MixColumns transformation to the state matrix."""
    for c in range(4):
        col_copy = []
        for r in range(4):
            col_copy.append(state[r][c])

        state[0][c] = (
            g_f_mult(INV_MIX_COL_MATRIX[0][0], col_copy[0]) ^
            g_f_mult(INV_MIX_COL_MATRIX[0][1], col_copy[1]) ^
            g_f_mult(INV_MIX_COL_MATRIX[0][2], col_copy[2]) ^
            g_f_mult(INV_MIX_COL_MATRIX[0][3], col_copy[3])
        )
        state[1][c] = (
            g_f_mult(INV_MIX_COL_MATRIX[1][0], col_copy[0]) ^
            g_f_mult(INV_MIX_COL_MATRIX[1][1], col_copy[1]) ^
            g_f_mult(INV_MIX_COL_MATRIX[1][2], col_copy[2]) ^
            g_f_mult(INV_MIX_COL_MATRIX[1][3], col_copy[3])
        )
        state[2][c] = (
            g_f_mult(INV_MIX_COL_MATRIX[2][0], col_copy[0]) ^
            g_f_mult(INV_MIX_COL_MATRIX[2][1], col_copy[1]) ^
            g_f_mult(INV_MIX_COL_MATRIX[2][2], col_copy[2]) ^
            g_f_mult(INV_MIX_COL_MATRIX[2][3], col_copy[3])
        )
        state[3][c] = (
            g_f_mult(INV_MIX_COL_MATRIX[3][0], col_copy[0]) ^
            g_f_mult(INV_MIX_COL_MATRIX[3][1], col_copy[1]) ^
            g_f_mult(INV_MIX_COL_MATRIX[3][2], col_copy[2]) ^
            g_f_mult(INV_MIX_COL_MATRIX[3][3], col_copy[3])
        )
    return state
```

# 8. Padding

```python
def pad(data_list):
    """Applies PKCS#7 padding to a list of bytes."""
    pad_len = 16 - (len(data_list) % 16)
    pad_byte_value = pad_len
    padding_list = [pad_byte_value] * pad_len
    return data_list + padding_list

def unpad(data_list):
    """
    Removes PKCS#7 padding from a list of decrypted bytes,
    verifying its full validity.
    """
    # 1. Protection against empty lists
    if not data_list:
        raise ValueError("Input data is empty.")

    # 2. Read the value of the last byte
    pad_len = data_list[-1]

    # 3. Preliminary range check
    if pad_len > 16 or pad_len < 1:
        raise ValueError(f"Invalid padding value: {pad_len}")

    # 4. Strict verification of all padding bytes
    # Check if all 'pad_len' of the last bytes have the value 'pad_len'
    for byte in data_list[-pad_len:]:
        if byte != pad_len:
            raise ValueError("PKCS#7 validation error: Inconsistent padding bytes.")

    # 5. Return data without padding
    return data_list[:-pad_len]
```

# 9.Main block cipher functions

```python
def aes_encrypt_block(plaintext_block, word_list):
    """Encrypts a single 16-byte plaintext block."""
    state = bytes_to_state(plaintext_block)

    # Initial Round (Round 0)
    state = add_round_key(state, word_list[0:4])

    # Main Rounds (Rounds 1 to 9)
    for r in range(1, NR):
        state = sub_bytes(state)
        state = shift_rows(state)
        state = mix_columns(state)
        state = add_round_key(state, word_list[r*4 : (r+1)*4])

    # Final Round (Round 10) - No MixColumns
    state = sub_bytes(state)
    state = shift_rows(state)
    state = add_round_key(state, word_list[NR*4 : (NR+1)*4])

    return state_to_bytes(state)


def aes_decrypt_block(ciphertext_block, word_list):
    """
    Decrypts a single 16-byte ciphertext block.

    Executes a full reverse cycle of 10 AES rounds, using
    inverse transformations and round keys in reverse order.

    Args:
        ciphertext_block (list): 16-byte encrypted block.
        word_list (list): Full list of 44 round key words generated by key_expansion.

    Returns:
        list: 16-byte decrypted block (plaintext).
    """
    state = bytes_to_state(ciphertext_block)

    # Initial Round (Reversing Round 10)
    state = add_round_key(state, word_list[NR*4 : (NR+1)*4])
    state = inv_shift_rows(state)
    state = inv_sub_bytes(state)

    # Main Rounds (Reversing Rounds 9 to 1)
    for r in range(NR - 1, 0, -1):
        state = add_round_key(state, word_list[r*4 : (r+1)*4])
        state = inv_mix_columns(state)
        state = inv_shift_rows(state)
        state = inv_sub_bytes(state)

    # Final Round (Reversing Round 0)
    state = add_round_key(state, word_list[0:4])

    return state_to_bytes(state)
```

# 10.Main file-handling functions

```python
def encrypt_data(data_bytes, main_key):
    """
    Encrypts arbitrary byte data using AES-128 in ECB mode.
    """
    if isinstance(main_key, bytes):
        main_key = list(main_key)

    word_list = key_expansion(main_key)
    padded_data = pad(list(data_bytes))
    encrypted_data = []

    for i in range(0, len(padded_data), 16):
        plaintext_block = padded_data[i : i+16]
        encrypted_block = aes_encrypt_block(plaintext_block, word_list)
        encrypted_data.extend(encrypted_block)

    return encrypted_data


def decrypt_data(encrypted_bytes, main_key):
    """
    Decrypts arbitrary byte data using AES-128 in ECB mode.
    """
    if isinstance(main_key, bytes):
        main_key = list(main_key)

    word_list = key_expansion(main_key)
    encrypted_list = list(encrypted_bytes)
    decrypted_data = []

    for i in range(0, len(encrypted_list), 16):
        ciphertext_block = encrypted_list[i : i+16]
        decrypted_block = aes_decrypt_block(ciphertext_block, word_list)
        decrypted_data.extend(decrypted_block)

    # Padding validation is handled inside the unpad function
    return unpad(decrypted_data)


def encrypt_file(in_filename, out_filename, main_key):
    """Reads a file, encrypts it, and writes the ciphertext to a new file."""
    try:
        with open(in_filename, 'rb') as f_in:
            file_data = f_in.read()
    except FileNotFoundError:
        print(f"Error: File '{in_filename}' not found.")
        return

    encrypted_data = encrypt_data(file_data, main_key)

    with open(out_filename, 'wb') as f_out:
        f_out.write(bytes(encrypted_data))


def decrypt_file(in_filename, out_filename, main_key):
    """Reads an encrypted file, decrypts it, and writes the plaintext to a new file."""
    try:
        with open(in_filename, 'rb') as f_in:
            encrypted_data = f_in.read()
    except FileNotFoundError:
        print(f"Error: File '{in_filename}' not found.")
        return

    if not encrypted_data:
        print("Error: Input file is empty.")
        return

    try:
        decrypted_data = decrypt_data(encrypted_data, main_key)
        with open(out_filename, 'wb') as f_out:
            f_out.write(bytes(decrypted_data))
    except ValueError as e:
        print(f"Decryption failed: {e}")
```

```python
from PIL import Image
import sys

INPUT_IMAGE = 'TEST_IMG.jpeg'
RAW_BIN_FILE = 'img_orig.bin'
ENC_BIN_FILE = 'img_enc.bin'
OUTPUT_IMAGE = 'img_enc.jpeg'

IMAGE_KEY = [
    0x2b, 0x7e, 0x15, 0x16, 0x28, 0xae, 0xd2, 0xa6,
    0xab, 0xf7, 0x15, 0x88, 0x09, 0xcf, 0x4f, 0x3c
]

if __name__ == "__main__":
    print("--- STARTING IMAGE ENCRYPTION DEMONSTRATION (ECB MODE) ---")

    # STEP 1: EXTRACT RAW PIXELS FROM IMAGE

    print("[STEP 1] Extracting raw RGB bytes from image...")
    try:
        with Image.open(INPUT_IMAGE) as img:
            # Ensure the image is in RGB format (no alpha channel)
            img = img.convert('RGB')
            img_size = img.size
            # tobytes() automatically and efficiently converts the image to raw bytes
            img_raw_bytes = img.tobytes()

        with open(RAW_BIN_FILE, 'wb') as f:
            f.write(img_raw_bytes)

        print(f"Extracted {len(img_raw_bytes)} bytes. Original size: {img_size}")

    except FileNotFoundError:
        print(f"ERROR: Image '{INPUT_IMAGE}' not found.")
        sys.exit(1)

    # STEP 2: ENCRYPT RAW BYTES

    print("\n[STEP 2] Encrypting raw image data...")
    # Korzystamy z naszej zrefaktoryzowanej funkcji
    encrypt_file(RAW_BIN_FILE, ENC_BIN_FILE, IMAGE_KEY)

    # STEP 3: RECONSTRUCT IMAGE FROM ENCRYPTED BYTES

    print("\n[STEP 3] Reconstructing image from encrypted bytes...")

    with open(ENC_BIN_FILE, "rb") as f:
        encrypted_bytes = f.read()

    # IMPORTANT: AES encryption added PKCS#7 padding.
    # To reconstruct the image, we need exactly the same number of bytes as the original.
    # Calculate the expected size (width * height * 3 RGB channels) and trim the padding.
    expected_size = img_size[0] * img_size[1] * 3
    encrypted_pixels = encrypted_bytes[:expected_size]

    # frombytes() instantly maps raw bytes back into a pixel matrix
    img_enc = Image.frombytes('RGB', img_size, encrypted_pixels)

    # Display and save the resulting image
    img_enc.show()
    img_enc.save(OUTPUT_IMAGE)
    print(f"\nSUCCESS: Encrypted image saved as '{OUTPUT_IMAGE}'.")
```
